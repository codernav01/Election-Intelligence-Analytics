## Missing Value Percentage
--- 

````python
def missing_value_report(dataset):

    report = pd.DataFrame({
        "Missing Values": dataset.isnull().sum(),
        "Missing Percentage": round((dataset.isnull().sum() / len(dataset)) * 100, 2)
    })

    return report


print("Election 2021")
display(missing_value_report(election_2021))

print("Election 2026")
display(missing_value_report(election_2026))

print("Constituency Master")
display(missing_value_report(constituency_master))
````

## Missing Value Treatment

Missing values are identified and treated according to business requirements.

Essential business columns such as Constituency, Candidate, Party, and Votes should not contain missing values.

Whitespace-only values are also treated as missing values.

## Convert Blank Strings to Missing Values
---

````python
datasets = [
    election_2021_clean,
    election_2026_clean,
    constituency_master_clean
]

for dataset in datasets:

    dataset.replace(r'^\s*$', np.nan, regex=True, inplace=True)
````

## Check Missing Values Again
---

````python
for name, dataset in {

    "Election 2021": election_2021_clean,

    "Election 2026": election_2026_clean,

    "Constituency Master": constituency_master_clean

}.items():

    print(f"\n{name}")

    display(dataset.isnull().sum())
````

````python
print("Rows Before Cleaning")

print(election_2021_clean.shape)

print(election_2026_clean.shape)

print(constituency_master_clean.shape)
````

````python
print("Rows After Duplicate Removal")

print(election_2021_clean.shape)

print(election_2026_clean.shape)

print(constituency_master_clean.shape)
````

## Whitespace Cleaning

Leading and trailing spaces often create duplicate categories during analysis.

All text columns are cleaned using string trimming.

````python
for dataset in datasets:

    text_columns = dataset.select_dtypes(include="object").columns

    dataset[text_columns] = dataset[text_columns].apply(lambda column: column.str.strip())
````

## 5.1 Create Working Copies
--- 

````python
election_2021_fe = election_2021_clean.copy()

election_2026_fe = election_2026_clean.copy()

constituency_master_fe = constituency_master_clean.copy()

print(election_2021_fe)
print(election_2026_fe)
print(constituency_master_fe)

print("Feature Engineering Working Copies Created Successfully")
````

## 5.2 Merge Constituency Master Data
---

````python
merge_columns = [
    "ac_number",
    "constituency",
    "district",
    "region",
    "reserved"
]

master = constituency_master_fe[merge_columns].drop_duplicates()

election_2021_fe = election_2021_fe.merge(
    master,
    on="ac_number",
    how="left",
    suffixes=("", "_master")
)

election_2026_fe = election_2026_fe.merge(
    master,
    on="ac_number",
    how="left",
    suffixes=("", "_master")
)

# ==========================================================
# Replace with Master Information
# ==========================================================

election_2021_fe["constituency"] = election_2021_fe["constituency_master"]
election_2021_fe["region"] = election_2021_fe["region_master"]
election_2021_fe["reserved"] = election_2021_fe["reserved_master"]

election_2021_fe.drop(
    columns=[
        "constituency_master",
        "region_master",
        "reserved_master"
    ],
    inplace=True,
    errors="ignore"
)

election_2026_fe["constituency"] = election_2026_fe["constituency_master"]
election_2026_fe["region"] = election_2026_fe["region_master"]
election_2026_fe["reserved"] = election_2026_fe["reserved_master"]

election_2026_fe.drop(
    columns=[
        "constituency_master",
        "region_master",
        "reserved_master"
    ],
    inplace=True,
    errors="ignore"
)

print("Master Data Successfully Merged")
````

## 5.3 Create Election Year Feature
---

````python
election_2021_fe["election_year"] = 2021

election_2026_fe["election_year"] = 2026

print("Election Year Feature Created")
````

## 5.4 Candidate Vote Percentage
---

````python
for dataset in [election_2021_fe, election_2026_fe]:

    total_votes = dataset.groupby("constituency")["votes"].transform("sum")

    dataset["vote_percentage"] = (
        dataset["votes"] / total_votes
    ) * 100

    dataset["vote_percentage"] = dataset["vote_percentage"].round(2)

print("Vote Percentage Feature Created")
````

````python
for dataset in [election_2021_fe, election_2026_fe]:

    dataset["total_constituency_votes"] = (
        dataset.groupby("constituency")["votes"]
        .transform("sum")
    )

print("Total Constituency Votes Feature Created")
````

````python
for dataset in [election_2021_fe, election_2026_fe]:

    dataset["total_candidates"] = (
        dataset.groupby("constituency")["candidate"]
        .transform("count")
    )

print("Total Candidates Feature Created")
````

## 5.5 Candidate Rank
--- 

````python
for dataset in [election_2021_fe, election_2026_fe]:

    dataset["candidate_rank"] = (
        dataset.groupby("constituency")["votes"]
        .rank(method="dense", ascending=False)
        .astype(int)
    )

print("Candidate Ranking Completed")
````

## 5.6 Winner Identification
--- 

````python
for dataset in [election_2021_fe, election_2026_fe]:

    dataset["winner"] = dataset["candidate_rank"].apply(
        lambda x: "Yes" if x == 1 else "No"
    )

print("Winner Feature Created")
````

## 5.7 Runner-Up Identification
---

````python
for dataset in [election_2021_fe, election_2026_fe]:

    dataset["runner_up"] = dataset["candidate_rank"].apply(
        lambda x: "Yes" if x == 2 else "No"
    )

print("Runner-Up Feature Created")
````

## 5.8 Winning Margin (Votes)
---

````python
def calculate_margin(df):

    top_votes = (
        df.groupby("constituency")["votes"]
        .nlargest(2)
        .reset_index(level=0)
    )

    margins = {}

    for constituency in df["constituency"].unique():

        values = (
            top_votes.loc[
                top_votes["constituency"] == constituency,
                "votes"
            ].tolist()
        )

        if len(values) >= 2:
            margins[constituency] = values[0] - values[1]
        else:
            margins[constituency] = values[0]

    df["winning_margin"] = df["constituency"].map(margins)

    return df

election_2021_fe = calculate_margin(election_2021_fe)

election_2026_fe = calculate_margin(election_2026_fe)

print("Winning Margin Feature Created")
````

## 5.9 Winning Margin Percentage
--- 

````python
for dataset in [election_2021_fe, election_2026_fe]:

    total_votes = dataset.groupby("constituency")["votes"].transform("sum")

    dataset["winning_margin_percentage"] = (
        dataset["winning_margin"] /
        total_votes
    ) * 100

    dataset["winning_margin_percentage"] = (
        dataset["winning_margin_percentage"]
        .round(2)
    )

print("Winning Margin Percentage Created")
````

## 5.10 Election Result Category
---

````python
def result_category(value):

    if value < 2:
        return "Very Close"

    elif value < 5:
        return "Close"

    elif value < 10:
        return "Competitive"

    elif value < 20:
        return "Comfortable"

    else:
        return "Landslide"

for dataset in [election_2021_fe, election_2026_fe]:

    dataset["result_category"] = (
        dataset["winning_margin_percentage"]
        .apply(result_category)
    )

print("Result Category Created")
````

## 5.11 Turnout Category
---

````python
def turnout_category(value):

    if pd.isna(value):
        return "Not Available"

    elif value < 60:
        return "Low"

    elif value < 70:
        return "Medium"

    elif value < 80:
        return "High"

    else:
        return "Very High"


for dataset in [election_2021_fe, election_2026_fe]:

    dataset["turnout_category"] = (
        dataset["turnout"]
        .apply(turnout_category)
    )

print("Turnout Category Created")
````

## 5.12 Reserved Seat Indicator
---

````python
for dataset in [election_2021_fe, election_2026_fe]:

    # Rename the columns created after merge
    if "reserved_y" in dataset.columns:
        dataset.rename(columns={"reserved_y": "reserved"}, inplace=True)

    if "region_y" in dataset.columns:
        dataset.rename(columns={"region_y": "region"}, inplace=True)

    # Remove duplicate columns
    dataset.drop(
        columns=["reserved_x", "region_x"],
        inplace=True,
        errors="ignore"
    )

print("Duplicate Columns Fixed Successfully")
````

````python
for dataset in [election_2021_fe, election_2026_fe]:

    dataset["reserved_indicator"] = (
        dataset["reserved"]
        .fillna("GENERAL")
        .astype(str)
        .str.upper()
        .apply(lambda x: 0 if x == "GENERAL" else 1)
    )

print("Reserved Indicator Created Successfully")
````

````python
vote_share = (

    election_2026_fe

    .groupby("party")["votes"]

    .sum()

    .reset_index()

)

total_votes = vote_share["votes"].sum()

vote_share["vote_share_percentage"] = (

    vote_share["votes"]

    /

    total_votes

) * 100

vote_share["vote_share_percentage"] = (

    vote_share["vote_share_percentage"]

    .round(2)

)

display(

    vote_share

    .sort_values(

        "vote_share_percentage",

        ascending=False

    )

    .head(10)

)
````

## Top Parties
--- 

````python
top_parties = (

    vote_share

    .sort_values(

        "vote_share_percentage",

        ascending=False

    )

    .head(10))

top_parties.plot(

    x="party",

    y="vote_share_percentage",

    kind="bar",

    legend=False,

    figsize=(10,6)

)

plt.title("Top Political Parties by Vote Share")
plt.xlabel("Political Party")
plt.ylabel("Vote Share (%)")
plt.xticks(rotation=45)
plt.show()
````
